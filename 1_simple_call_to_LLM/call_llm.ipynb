```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

response = llm.invoke("biggest country in the wold?")
print(response.text)
```

To give instructions

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke([
    ["system","you are an helpful assistant that answers in short one line"],
    ["human" ,"what is the biggest country in the world"]
])
print(response.text)
```

change creativity

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

response = llm.invoke("write a first sentence for IT job post application email")
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=1)

response = llm.invoke("write a first sentence for IT job post application email")
print(response.text)
```

__ Call llms using Groq models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3-32b" , reasoning_format="parsed")

messages = [
    (
        "system",
        "You are helpful assistant that gives a short answer to a question"
    ),
    (
        "human",
        "How much longer cow lives?"
    ),
]
ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
